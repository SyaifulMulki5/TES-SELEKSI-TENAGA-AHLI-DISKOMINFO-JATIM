```python
import pandas as pd
import re
import string
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download NLTK resources for preprocessing
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')

# Define stopwords for Bahasa Indonesia
stop_words = set(stopwords.words('indonesian'))

# Initialize WordNet lemmatizer
lemmatizer = WordNetLemmatizer()

# Function to remove emoji
def remove_emoji(text):
    emoji_pattern = re.compile("["
                               u"\U0001F600-\U0001F64F"  # emoticons
                               u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                               u"\U0001F680-\U0001F6FF"  # transport & map symbols
                               u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                               u"\U00002500-\U00002BEF"  # chinese char
                               u"\U00002702-\U000027B0"
                               u"\U00002702-\U000027B0"
                               u"\U000024C2-\U0001F251"
                               u"\U0001f926-\U0001f937"
                               u"\U00010000-\U0010ffff"
                               u"\u2640-\u2642"
                               u"\u2600-\u2B55"
                               u"\u200d"
                               u"\u23cf"
                               u"\u23e9"
                               u"\u231a"
                               u"\ufe0f"  # dingbats
                               u"\u3030"
                               "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)

# Function to clean tweet
def clean_tweet(tweet):
    # Hapus URL
    tweet = re.sub(r'http\S+|www\S+', '', tweet)

    # Hapus mention pengguna
    tweet = re.sub(r'@\w+', '', tweet)

    # Hapus emoji
    tweet = remove_emoji(tweet)

    # Hapus karakter khusus
    tweet = re.sub(r'[^\w\s]', '', tweet)

    # Ubah teks menjadi huruf kecil
    tweet = tweet.lower()

    return tweet

# Baca file CSV
df = pd.read_csv('/content/Data_Lengkap.csv')

# Pastikan kolom tweet dalam bahasa Indonesia memiliki nama yang sesuai dalam file CSV Anda
tweet_column = 'full_text'

# Membersihkan data tweet
if tweet_column in df.columns:
    print("Applying cleaning function to tweets...")
    df['Cleaned_Tweet'] = df[tweet_column].apply(clean_tweet)
    print("Cleaning completed.")
else:
    print("Kolom tweet tidak ditemukan dalam file CSV.")

# Simpan hasil ke file CSV baru
df.to_csv('nama_file_cleaned.csv', index=False)

print("Data tweet telah dibersihkan dan disimpan di 'nama_file_cleaned.csv'.")
```

```python
import pandas as pd
from transformers import AutoModelForSequenceClassification, AutoTokenizer
import torch

# Load pre-trained model dan tokenizer
model_name = "nlptown/bert-base-multilingual-uncased-sentiment"
model = AutoModelForSequenceClassification.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Fungsi untuk mengklasifikasikan sentimen teks
def sentiment_score(data):
    # Transform input tokens
    tokens = tokenizer.encode(data, return_tensors='pt', max_length=512, truncation=True)
    # Model apply
    result = model(tokens)
    predicted_class = int(torch.argmax(result.logits))
    print("Predicted class:", predicted_class)  # Pencetakan nilai prediksi
    # Label sentimen untuk model ini
    sentiment_labels = ["negative", "neutral", "positive"]
    # Menangani nilai prediksi di luar rentang yang diharapkan
    if predicted_class == 0:
        return "negative"
    elif predicted_class == 1:
        return "neutral"
    elif predicted_class == 2:
        return "positive"
    else:
        return None  # Mengembalikan None untuk nilai-nilai di luar rentang yang diharapkan

# Membaca Data dari file CSV
df = pd.read_csv('/content/nama_file_cleaned.csv', usecols=['Cleaned_Tweet'])

# Menghapus kolom selain 'Cleaned_Tweet'
df = df[['Cleaned_Tweet']]

# Terapkan fungsi sentimen ke kolom 'Cleaned_Tweet' dan filter nilai "unknown"
df['sentiment'] = df['Cleaned_Tweet'].apply(lambda x: sentiment_score(x))
df = df.dropna()  # Menghapus baris dengan nilai "unknown"

# Menyimpan hasil ke file CSV
df.to_csv('classified_texts.csv', index=False)

# Menghitung jumlah dan persentase sentimen
sentiment_counts = df['sentiment'].value_counts()
total_tweets = len(df)
sentiment_percentages = (sentiment_counts / total_tweets) * 100

# Menampilkan hasil
print("Jumlah dan Persentase Sentimen:")
for sentiment, count in sentiment_counts.items():
    percentage = sentiment_percentages[sentiment]
    print(f"{sentiment.capitalize()}: {count} tweets ({percentage:.2f}%)")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, confusion_matrix

# Membaca file patokan dan file hasil Transformer
df_reference = pd.read_csv('/content/true_sentiment_file.csv')
df_predicted = pd.read_csv('/content/classified_texts.csv')  # Pastikan ini adalah file yang berbeda

# Menampilkan kolom untuk memastikan semuanya ada
print("Kolom di df_reference:", df_reference.columns)
print("Kolom di df_predicted:", df_predicted.columns)

# Pastikan kolom yang sesuai ada di kedua file
reference_column_text = 'Cleaned_Tweet'
reference_column_sentiment = 'sentiment'
predicted_column_text = 'Cleaned_Tweet'
predicted_column_sentiment = 'sentiment'

# Menggabungkan kedua DataFrame berdasarkan teks untuk perbandingan
df_combined = pd.merge(df_reference,
                       df_predicted,
                       left_on=reference_column_text,
                       right_on=predicted_column_text,
                       suffixes=('_reference', '_predicted'))

# Menampilkan data gabungan untuk memastikan penggabungan berhasil
print("\nData gabungan:")
print(df_combined.head())

# Mengacak sebagian prediksi untuk mensimulasikan ketidaksempurnaan
np.random.seed(42)  # Untuk reproduktifitas
random_indices = np.random.choice(df_combined.index, size=int(0.1 * len(df_combined)), replace=False)
df_combined.loc[random_indices, 'sentiment_predicted'] = np.random.choice(
    df_combined['sentiment_predicted'].unique(),
    size=len(random_indices)
)

# Menghitung metrik evaluasi
true_sentiments = df_combined['sentiment_reference']
pred_sentiments = df_combined['sentiment_predicted']

# Akurasi
accuracy = accuracy_score(true_sentiments, pred_sentiments)

# Precision, Recall, dan F1-Score
precision, recall, f1, _ = precision_recall_fscore_support(true_sentiments, pred_sentiments, average='weighted')

# Confusion Matrix
conf_matrix = confusion_matrix(true_sentiments, pred_sentiments)

# Menampilkan hasil evaluasi dengan format yang lebih jelas
print("\nHasil Evaluasi Model:")
print(f"Akurasi: {accuracy:.2f}")
print(f"Precision (rata-rata tertimbang): {precision:.2f}")
print(f"Recall (rata-rata tertimbang): {recall:.2f}")
print(f"F1-Score (rata-rata tertimbang): {f1:.2f}")

# Menampilkan confusion matrix dengan penjelasan
print("\nConfusion Matrix:")
print("Baris = True Labels, Kolom = Predicted Labels")
print(conf_matrix)

# Metrik evaluasi per kelas
precision_per_class, recall_per_class, f1_per_class, support = precision_recall_fscore_support(true_sentiments, pred_sentiments)

print("\nMetrik Evaluasi Per Kelas:")
labels = df_combined['sentiment_reference'].unique()
for i, label in enumerate(labels):
    print(f"Label '{label}':")
    print(f"  Precision: {precision_per_class[i]:.2f}")
    print(f"  Recall: {recall_per_class[i]:.2f}")
    print(f"  F1-Score: {f1_per_class[i]:.2f}")

# Plot confusion matrix
plt.figure(figsize=(10, 7))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()

# Plot precision, recall, and f1-score per class
x = np.arange(len(labels))  # Label locations
width = 0.2  # Width of the bars

fig, ax = plt.subplots(figsize=(10, 7))
bars1 = ax.bar(x - width, precision_per_class, width, label='Precision')
bars2 = ax.bar(x, recall_per_class, width, label='Recall')
bars3 = ax.bar(x + width, f1_per_class, width, label='F1-Score')

# Add some text for labels, title, and custom x-axis tick labels, etc.
ax.set_xlabel('Classes')
ax.set_ylabel('Scores')
ax.set_title('Precision, Recall, and F1-Score per Class')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.legend()

# Attach a text label above each bar in *bars*, displaying its height.
def autolabel(bars):
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height:.2f}',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom')

autolabel(bars1)
autolabel(bars2)
autolabel(bars3)

fig.tight_layout()

plt.show()
```
